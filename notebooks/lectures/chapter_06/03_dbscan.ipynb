# Capítulo 6 · DBSCAN con dos lunas

¿Cuándo consigue DBSCAN que cada luna quede en un grupo distinto? Vamos a probar varias combinaciones sobre los mismos 200 puntos. Cambiaremos dos valores.

- **Radio (`eps`)**. Indica hasta qué distancia llega el círculo alrededor de cada punto.
- **Mínimo de puntos (`min_samples`)**. Indica cuántos puntos debe contener ese círculo, contando el del centro, para que el centro sea un **núcleo**.

DBSCAN conecta los núcleos cercanos y añade los puntos que quedan a su alcance. Los puntos que quedan fuera de todos los grupos se marcan como **ruido**.

```python
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons

X_lunas, _ = make_moons(n_samples=200, noise=0.08, random_state=42)
```

## Probar seis combinaciones

Cruzamos tres radios con dos mínimos. Lee la figura **de izquierda a derecha** para aumentar el radio manteniendo el mismo mínimo. Lee **de arriba abajo** para aumentar el mínimo con el radio fijo. En cada panel, los colores distinguen los grupos y las cruces grises indican ruido.

```python
radios = [0.16, 0.20, 0.34]
minimos = [5, 8]
```

```python
fig, ejes = plt.subplots(len(minimos), len(radios), figsize=(11, 5.5),
                         sharex=True, sharey=True, squeeze=False, layout="constrained")

for fila, minimo in enumerate(minimos):
    for columna, radio in enumerate(radios):
        modelo = DBSCAN(eps=radio, min_samples=minimo)
        etiquetas = modelo.fit_predict(X_lunas)
        ruido = etiquetas == -1
        grupos = set(etiquetas[~ruido])

        eje = ejes[fila, columna]
        eje.scatter(X_lunas[~ruido, 0], X_lunas[~ruido, 1], c=etiquetas[~ruido], cmap="tab10", s=16)
        eje.scatter(X_lunas[ruido, 0], X_lunas[ruido, 1], color="gray", marker="x", s=22)
        eje.set_title(f"Radio {radio:.2f} · mínimo {minimo}\nGrupos: {len(grupos)} · Ruido: {ruido.sum()}")
        eje.set_aspect("equal")

plt.show()
```

**¿Qué se ve?** Recorre la fila del mínimo de 5. Con radio 0.16, una misma luna aparece partida en varios colores. Con 0.20, cada arco tiene su propio color. Con 0.34, un puente de puntos cercanos une las dos lunas en un solo grupo.

Ahora mira la columna del radio 0.16. Al subir el mínimo de 5 a 8, los puntos marcados como ruido pasan de 7 a 26. El círculo sigue teniendo el mismo tamaño, pero en los tramos menos densos cuesta más reunir ocho puntos.
