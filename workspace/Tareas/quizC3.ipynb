```python
#EL MAYOR PESO ENTRE LAS ESTUDIANTES DE ESTATURA INFERIROR A 1.60 M

estudiantes_f = students["Gender"] == "F" #la mascara para solo contar a las estudiantes

alturas = students["Height"] < 160

peso_mayor_estudiantes = students.loc[estudiantes_f & alturas, "Weight"]

print("El peso mayor entre las estudiantes de menos de 1.60 m es: ",peso_mayor_estudiantes.max())
```

```python
#ALTURA PROMEDIO DE LOS ALUMNOS CON GRUPO SANGUINEO 'A' o 'AB'

tipo_sang_a = students["Bloodtype"] == "A"
tipo_sang_ab = students["Bloodtype"] == "AB"  

alt_prom = students.loc[tipo_sang_a | tipo_sang_ab, "Height"]

print("La altura promedio de los alumnos con tipo sangguineo 'A' o 'AB' es: ",alt_prom.mean())
```

1) Genera una tabla de frecuencias para la variable 'yearOfStart' y visualízala por año.
•	Ordena por año.
•	Dibuja una gráfica de línea de 1997 a 2014. <= Pista: plt.xlim()


```python
frecuencia = coffee["Year of Start"].value_counts().sort_index()
print(frecuencia)

fig, ax = plt.subplots(figsize = (10,5))
ax.plot(frecuencia.index, frecuencia.values, marker="o")
ax.set_xlim(1997, 2014)

ax.set_xlabel("Año de inicio")
ax.set_ylabel("Número de cafeterías")
plt.show()

```

2) Divide los datos por el estado actual del negocio ('In' o 'Out'). Después, visualiza la tendencia annual de las frecuencias de la variable 'yearOfStart'.
•	Ordena por año.
•	Dibuja dos gráficas de líneas superpuestas de 1997 a 2014.
•	Usa el objeto 'figure'.


```python
frecuencia_in = (coffee.loc[coffee["Current State"] == "In", "Year of Start"].value_counts().sort_index())

frecuencia_out = (coffee.loc[coffee["Current State"] == "Out", "Year of Start"].value_counts().sort_index())

years = range(1997, 2015)

frecuencia_in = frecuencia_in.reindex(years, fill_value=0)
frecuencia_out = frecuencia_out.reindex(years, fill_value=0)

fig = plt.figure(figsize=(10, 4))
ax = fig.add_subplot(111)
ax.plot(frecuencia_in.index, frecuencia_in.values, marker="o", label="In")
ax.plot(frecuencia_out.index, frecuencia_out.values, marker="o", label="Out")
ax.set_xlim(1997, 2014)
ax.set_xlabel("Año de inicio")
ax.set_ylabel("Número de cafeterías")
ax.legend()
plt.show()
```
